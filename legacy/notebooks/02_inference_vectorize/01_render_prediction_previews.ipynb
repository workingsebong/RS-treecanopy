# 01. Render Prediction Previews

Run the trained YOLO-seg model on the test tiles and render orange tree crown mask previews.

**Input**
- `outputs/yolo/tree_crown_512_yolo11n_seg_e50/weights/best.pt`
- `data/processed/yolo_seg_512_random_seoul_urban/images/test/`

**Output**
- `outputs/yolo/tree_crown_512_yolo11n_seg_e50_test_predict_orange/`

```python
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve()
while PROJECT_ROOT != PROJECT_ROOT.parent and not (PROJECT_ROOT / "tools").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

PROJECT_ROOT
```

```python
import subprocess

cmd = ["python", "tools/render_yolo_seg_previews.py"]
subprocess.run(cmd, cwd=PROJECT_ROOT, check=True)
```

```python
preview_dir = PROJECT_ROOT / "outputs/yolo/tree_crown_512_yolo11n_seg_e50_test_predict_orange"
preview_files = sorted(preview_dir.glob("*.jpg"))
len(preview_files), preview_files[:3]
```

```python
from IPython.display import Image, display

for path in preview_files[:3]:
    print(path.name)
    display(Image(filename=str(path)))
```
