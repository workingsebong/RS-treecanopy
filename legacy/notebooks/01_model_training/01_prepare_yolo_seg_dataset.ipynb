# 01. Prepare YOLO Segmentation Dataset

Convert labeling-app tree crown polygons into a YOLO segmentation dataset.

**Input**
- `data/processed/labeling_candidates_512_random_seoul_urban/images/`
- `data/processed/labels_512_random_seoul_urban/{json,yolo_seg,geojson}/`

**Output**
- `data/processed/yolo_seg_512_random_seoul_urban/`

The train/val/test split is patch-separated, not administrative district-separated.

```python
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve()
while PROJECT_ROOT != PROJECT_ROOT.parent and not (PROJECT_ROOT / "tools").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

PROJECT_ROOT
```

```python
import subprocess

cmd = ["python", "tools/prepare_yolo_seg_dataset.py", "--overwrite"]
subprocess.run(cmd, cwd=PROJECT_ROOT, check=True)
```

```python
import yaml

dataset_dir = PROJECT_ROOT / "data/processed/yolo_seg_512_random_seoul_urban"
dataset_yaml = dataset_dir / "dataset.yaml"

with dataset_yaml.open("r", encoding="utf-8") as f:
    dataset_config = yaml.safe_load(f)

dataset_config
```

```python
for split in ["train", "val", "test"]:
    image_count = len(list((dataset_dir / "images" / split).glob("*.png")))
    label_count = len(list((dataset_dir / "labels" / split).glob("*.txt")))
    print(f"{split}: {image_count} images, {label_count} labels")
```
