# 02. Train YOLO11n Segmentation Baseline

Train a small YOLO11n-seg baseline on the 512px random Seoul urban tree crown dataset.

**Input**
- `data/processed/yolo_seg_512_random_seoul_urban/dataset.yaml`

**Output**
- `outputs/yolo/tree_crown_512_yolo11n_seg_e50/`
- `outputs/yolo/tree_crown_512_yolo11n_seg_e50_test/`

Current pilot result: test mask mAP50 ≈ 0.353, mask mAP50-95 ≈ 0.134.

```python
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve()
while PROJECT_ROOT != PROJECT_ROOT.parent and not (PROJECT_ROOT / "tools").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_YAML = PROJECT_ROOT / "data/processed/yolo_seg_512_random_seoul_urban/dataset.yaml"
OUTPUT_ROOT = PROJECT_ROOT / "outputs/yolo"
RUN_NAME = "tree_crown_512_yolo11n_seg_e50"

OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)
DATA_YAML
```

```python
from ultralytics import YOLO

model = YOLO("yolo11n-seg.pt")
train_results = model.train(
    data=str(DATA_YAML),
    epochs=50,
    imgsz=512,
    batch=2,
    device="cpu",  # Change to 0 when CUDA is available.
    workers=0,
    project=str(OUTPUT_ROOT),
    name=RUN_NAME,
    exist_ok=True,
    patience=20,
)
```

```python
best_weight = OUTPUT_ROOT / RUN_NAME / "weights/best.pt"
best_weight
```

```python
best_model = YOLO(str(best_weight))
test_metrics = best_model.val(
    data=str(DATA_YAML),
    split="test",
    imgsz=512,
    batch=2,
    device="cpu",
    workers=0,
    project=str(OUTPUT_ROOT),
    name=f"{RUN_NAME}_test",
    exist_ok=True,
)
test_metrics
```
